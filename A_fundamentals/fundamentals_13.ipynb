## 1. Dictionaries
----------
Python dictionaries are a powerful and versatile data structure used to store and manage collections of key-value pairs. 

Each key in a dictionary is unique and associated with a value. 

Dictionaries are often referred to as "associative arrays" or "hash maps" in other programming languages. 

Below is an introduction to dictionaries with some examples.

## 1.1 Creating a Dictionary:
You can create a dictionary by enclosing a comma-separated list of key-value pairs within curly brackets {}.

```python
# Creating a dictionary
student = {
    "name": "Munazza",
    "age": 20,
    "major": "Computer Science"
}
print(student)
```

## 1.2 Accessing Values:
You can access the values in a dictionary by referencing the keys

```python
# Accessing values
print(student["name"])    # Output: Munazza
print(student["age"])     # Output: 20
print(student["major"])   # Output: Computer Science
```

## 1.3 Adding and Modifying Entries:
You can add new entries to a dictionary or modify existing ones using the keys.

```python
# Adding a new entry
student["gender"] = "Female"
print(student)

# Modifying an entry
student["age"] = 21
print(student)
```

## 1.4 Dictionary Methods:
Dictionaries have several methods that allow you to manipulate and retrieve information. The [python documentation](https://docs.python.org/3/library/stdtypes.html#dictionary-view-objects) has a complete list of the methods available for dictionaries.

```python
# Getting all keys and values
keys = student.keys()
values = student.values()
print('keys=', keys)
print('values=', values)

# Checking if a key exists
if "name" in student:
    print("Name exists in the dictionary")

# Removing an entry
removed_value = student.pop("age")
print(student)
```

## 1.5 Iterating Through a Dictionary:
You can iterate through the keys or key-value pairs of a dictionary.

```python
# Iterating through keys
# (NB writing "for key in student" gives identical results to ""for key in student.keys() )
print('\n Iterating through keys ...')
for key in student:
    print(key, student[key])

# Iterating through key-value pairs
print('\n Iterating through key-value pairs ...')
for key, value in student.items():
    print(key, value)
```

## 1.6 Use Cases:
Dictionaries are handy in various scenarios, such as:
1. Mapping: Creating relationships between keys and corresponding values, such as:
    - Data Storage: Storing data with meaningful labels, like storing student information.
    - Configuration Settings: Storing configuration settings with associated values.
2. Counting Occurrences: Counting the occurrences of elements in a dataset. See example below.

```python
word_counts = {}
text = "This is The sample text to count word occurrences. It contains multiple words. Some of the words occur multiple times within the text."
words = text.split()

for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

for key in sorted(list(word_counts.keys())):
    print(f'The word "{key}" occurs {word_counts[key]} times')
```
