# 1: Basic Error Types

## 1.1 TypeError

In the following example, a `TypeError` is raised because we are trying to perform an operation (squaring) that cannot be done on this *type* of variable ( a *string* in the example below). 

```python
# Create a string variable 
s = 'aa'
# Try to raise `s` to the power of 2
r = s**2 
```

Note that in the example above, 
1. the debugger points out the part of the code that is going wrong. 
2. the debugger tells you the type of problem (TypeError), as well as supplying more info:
 - *unsupported operand type(s) for ** or pow(): 'str' and 'int'*
3. The error is saying you're trying to exponentiate a *string* with an *int* (integer) when you need to be using two integer values. 

This is to help you find and fix your problem code!


## 1.2 ValueError

In the following example, a `ValueError` is raised because we are trying to perform an operation (reshaping a numpy array) that cannot be done using the numerical values supplied.

```python
# Create a 1D array with 10 elements
arr = np.arange(10)
print('original array', arr)

# Attempt to reshape it into a shape that's incompatible with its total number of elements
# This will raise a ValueError because 3*4 = 12 != 10
reshaped_arr = arr.reshape((3, 4))
```

In the example above, 
1. the debugger points out the part of the code that is going wrong. 
2. the debugger tells you the type of problem (TypeError), as well as supplying more info:
 - *cannot reshape array of size 10 into shape (3,4)*

```python
# this is more appropriate and does not raise an error
reshaped_arr = arr.reshape((2, 5))
print('reshaped array\n', reshaped_arr)
```

## 1.3 Syntax Error
In the following example, a `SyntaxError` is raised because we have written invalid code.

```python
if x = 1:
    print("x is 1")
```

In the example above, 
1. the debugger points out the part of the code that is going wrong. 
2. the debugger tells you the type of problem (SyntaxError), as well as supplying more info:
 - *invalid syntax*
 - In this case it should have been "==" and not "="

## 1.4 NameError

In the following example, a `NameError` is raised because we use a variable without having first declared it. 

```python
print(undefined_variable)
```

Again, in the example above, 
1. the debugger points out the part of the code that is going wrong. 
2. the debugger tells you the type of problem (NameError), as well as supplying more info:
 - *name 'undefined_variable' is not defined*

# 2. Debugging Errors in Modules 

Let's look at the errors that are generated when we use code that is contained in a module that we are writing and importing.

## 2.1 Code Works, Poor Input ... 

We will start by writing some simple functions to an external module.

We will then import the module and call the functions with some differing input. 

```python
%%file fundamentals_14_funcs.py

def add_and_square(a,b):
    """ A function to add to numbers together and then square the sum"""
    return (a+b)**2
def repeated_call(a,b,c,d):
    """ A function that calls *add_and_square* on 2 pairs of numbers """
    r1 = add_and_square(a,b)
    r2 = add_and_square(c,d)
    r3 = add_and_square(r1,r2)
    return r3
    
```

#### We start by calling the function using acceptable values for the input variables

```python
import fundamentals_14_funcs as f14  
# Define four variables: all are integers
a,b,c,d = 2,4,6,8
# Call the function named, `repeated_call`
r = f14.repeated_call(a,b,c,d)
print(r)
```

#### Now we call the function with inappropriate (string) values for some of the input variables:

```python
import fundamentals_14_funcs as f14 
# Define four variables: all are integers
a,b,c,d = 2,4,6,'vkbjasd'
# Call the function named, `repeated_call`
r = f14.repeated_call(a,b,c,d)
print(r)
```

Note that the returned debugging information is now more complicated.
 - ***The debugger guides us through the code and highlights where the problem is. ***

In the first block of debugging output, it tells us which part of the notebook cell went wrong ...
 - `----> 5 r = f14.repeated_call(a,b,c,d)`
 
Then in the second block of debugging output, it digs further into the code and tells us which part of `f14.repeated_call` went wrong ...
 - `----> 8     r2 = add_and_square(c,d)`
 
Then in the third block of debugging output, it digs even further into the code and tells us which part of `add_and_square` went wrong ...
 - `----> 4     return (a+b)**2` 
 
And then finally, as seen previously, we get the details of the actual error that was encountered ...
 - `TypeError:`
 - `unsupported operand type(s) for +: 'int' and 'str'`
 - `you can't add together an integer and a string (they have to be of the same type)`

## 2.2 Bad Code

In the example above, the code itself functioned if given correct inputs. 

In the example below, the code will never work as it is (incorrectly) written. 

Let's see what happens when we save the code to file, import it and execute it ...

```python
%%file fundamentals_14_badcode.py

def badly_written_func(a,b):
    """ A badly written function what should generate a *KeyError* """
    d = {} 
    return d[a] + d[b]

def repeat_calls_to_bad_func(a,b,c,d):
    """ A function that calls *badly_written_func* on sets of input variables """
    r1 = badly_written_func(a,b)
    r2 = badly_written_func(c,d)
    r3 = badly_written_func(r1,r2)
    return r3
    
```

```python
# import the new (bad) code 
import fundamentals_14_badcode as f14bad 
from imp import reload 
reload(f14bad)
# Define four variables: all are integers
a,b,c,d = 2,4,6,8
# Call the function, `repeat_calls_to_bad_func`
r = f14bad.repeat_calls_to_bad_func(a,b,c,d)
print(r)
```

Similar to the example in Section 2.1, the returned debugging information is lengthy.
 - ***Again, the debugger is guiding us through the code and highlighting where the problems are. ***

In the first block of debugging output, it tells us which part of the notebook cell went wrong ...
 - `----> 7 r = f14bad.repeat_calls_to_bad_func(a,b,c,d)`
 
Then in the second block of debugging output, it digs further into the code and tells us which part of `f14bad.repeat_calls_to_bad_func` went wrong ...
 - `----> 9     r1 = badly_written_func(a,b)`
 
Then in the third block of debugging output, it digs even further into the code and tells us which part of `badly_written_func` went wrong ...
 - `----> 5     return d[a] + d[b]` 
 
And then finally, as seen previously, we get the details of the actual error that was encountered ...
 - `KeyError: 2`

Note: 
 - In the code above there are actually two errors 
    - `d[a]` <<-- `a` is not a key in the dictionary `d`
    - `d[b]` <<-- `b` is not a key in the dictionary `d`
 - But the debugger only tells us about the first one it comes across. 
    - `KeyError: 2` <<-- I.e. the variable `a=2` is not a key in the dictionary `d`
 - If we fix *only* the first error (e.g., we might change `return d[a] + d[b]` to ... `return a + d[b]`), then when we call the function again, the debugger will tell us about the remaining error ...
   - `KeyError: 4`
