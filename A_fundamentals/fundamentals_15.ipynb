## 1. Reading & Writing: Basic Built-In Methods

### 1.1 Writing

We will first write to files using the standard built-in method. 

We use the `open` command to create a *file-object*. The file-object is given the variable name *fo* in the code below. 

The file-object can then be used to perform read and write operations. 

```python
# Create a list of data that we want to print to file 
lst = ['sdgsdGdg','34234','abfa','1.2']

# Define the filename we want to create
filename = 'basic_output_file.txt'

# Open the file 
# Note: We provide the 'w' option here to indicate that we want to WRITE to the file 
with open(filename, 'w') as fo:
    # Loop over the lines in the list 
    for line in lst:
        # Write each line to the file 
        # Note 1 the use of the 'write' function
        # Note 2 the need to add a '\n' to write a new line to the file for each item
        fo.write(line + '\n')

# The file should now contain content that looks like the 4 lines below (but not the """ lines)
# The file name 'basic_output_file.txt' should be in the same directory as this notebook: go and check!!
"""
sdgsdGdg
34234
abfa
1.2
"""
```

### 1.2 Reading

Reading from a file using the standard built-in method. 


```python
# Define the filename we want to read from
filename = 'basic_output_file.txt'

# Open the file 
# Note: We use the 'r' option to specify that we want to READ from the file 
with open(filename, 'r') as fo:
    # Loop over the lines in the file 
    # Note 1: The file object, 'fo', can be iterated over 
    # Note 2: The new line, '\n', that is present at the end of each line in the file is explicitly printed-out
    for line in fo:
        print(line)
```

Populating a list of data from a file using the standard built-in method. 


```python
# Define the filename we want to read from
filename = 'basic_output_file.txt'

# Open the file 
# Note: We use the 'r' option to specify that we want to READ from the file 
with open(filename, 'r') as fo:
    # Create a list that we will add the data to 
    data_list = [] 
    
    # Loop over the lines in the file 
    for line in fo:
        # append the lines of data into your list
        # need to strip off the 'new line' at the end of the line
        data_list.append( line.strip() ) 
#        data_list.append( line ) 
print(data_list)
```

## 2.1 Writing to file using 'savetxt'

```python
import numpy as np 

# Create an array 
arr = np.arange(1,10,1)
print(arr) 

# Save the data to file using the numpy 'savetxt' function
filename = 'numpy_example.txt'
np.savetxt(filename, arr, delimiter=',')
```

## 2.2 Reading from file using 'loadtxt'

```python
# Read from file using 'loadtxt'
filename = 'numpy_example.txt'
read_arr = np.loadtxt(filename, delimiter=',')
print(read_arr)
```

# 3. Reading and Writing Dictionaries as JSON files 
Dictionaries can be directly saved to (and read from) JSON files (see https://en.wikipedia.org/wiki/JSON). 

## 3.1 Writing a dictionary to file

```python
# Create a sample dictionary 
d = { 'first_name' : 'Neil', 
    'last_name' : 'deGrasse Tyson' , 
    'pet': 'dog', 
    'something': 'buvhasdfjkvhsdf',
    'lst': [1, 2.0, '3.00'], 
    'another_dict': {'k1':'v1' , 
                     'k2':'v2'}}

# Import json package and use 'dump' routine to save to file 
# NB(1): 
import json
filename = 'fundamentals_15_sample.json'
with open(filename, 'w', encoding='utf-8') as fo:
    json.dump(d, fo, ensure_ascii=False, indent=4)
    
```

Check that the data appear in the file "fundamentals_15_sample.json" in the same directory as this notebook. 


## 3.2 Read dictionary from file 

```python
# Import json package 
import json

# define the file to be read
filename = 'fundamentals_15_sample.json'

# open the file 
with open(filename) as fo:
    # use the 'load' routine to read from file 
    data_loaded = json.load(fo) 
    
# print out the keys and values    
for k,v in data_loaded.items():
    print(k,v)
```

# 4. Reading and Writing data to/from Pandas Dataframes

The "pandas" package is widely used for data analysis. 

It is designed for working with "tabular" data. 

It also contains a number of convenient functions for reading and writing from/to file. 

***We will learn more about the pandas package in the data-analysis section of the course,*** but we'll introduce its data reading and writing capabilities here.  

## 4.1 Writing a dataframe to a csv

```python
# Ensure pandas has been imported 
import pandas as pd 

# Create Python lists
names = ["Alice", "Bindu", "Charlie", "Diego"]
ages = [25, 70, 58, 12]

# Create a dictionary and then use this to create a Pandas DataFrame
data = {
    "Name": names,
    "Age": ages
}
df = pd.DataFrame(data)

# Display the DataFrame
print(df)

# Save the dataframe to file using the 'to_csv' function 
# Note the use of the ", index=False" option to prevent the index field being printed to file. 
#    - Trying commenting out this option and rerunning the routine: what does the file look like? 
filename = 'ex4_dataframe.csv'
df.to_csv(filename, index=False)
```

## 4.2 Read a csv into a dataframe 

```python
# Use the file we created above to create a new pandas dataframe
filename = 'ex4_dataframe.csv'
new_df = pd.read_csv(filename)

# Display the DataFrame
print(new_df)
```
