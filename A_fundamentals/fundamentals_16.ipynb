# 1. NumPy

We have used NumPy several times already, starting with our introduction to arrays in [Fundamentals 4](fundamentals_4.ipynb). 

The main purpose of the NumPy package is to provide a powerful and efficient way to work with numerical data in Python. 

Some key functionalities:
 - Arrays and Matrices: NumPy's core feature is the ndarray (n-dimensional array) object. It allows you to efficiently store and manipulate large datasets of homogeneous data. You can create arrays with various dimensions (1D, 2D, 3D, and more) and perform mathematical operations on them.
 - Mathematical Functions: NumPy provides a wide range of mathematical functions that operate element-wise on arrays. These functions are optimized for performance and allow you to perform complex operations on large datasets easily.
 - Linear Algebra: NumPy has a comprehensive linear algebra module (numpy.linalg) that provides functions for matrix operations, eigenvalue decomposition, singular value decomposition, and more.
 - Efficiency: NumPy is implemented in C and optimized for performance. It's much faster than using standard Python lists for numerical operations, making it suitable for handling large datasets and complex calculations.
 - Integration with Other Libraries: NumPy is the foundation for many other libraries and tools used in data science, machine learning, scientific computing, and more. It provides an essential toolkit for working with numerical data efficiently and effectively in Python.

# 2. Pandas

In [Fundamentals 15](fundamentals_15.ipynb), we briefly looked at using the pandas package in the context of reading and writing csv files. 

The pandas package in Python is a widely used library for data manipulation, analysis, and manipulation. It provides powerful and flexible data structures for effectively working with structured data and time series data. Pandas is designed to make data analysis tasks easier and more efficient, especially for tasks related to tabular and time-series data. 

Some key functionalities:
1. Data Structures: Pandas uses two primary data structures:

3. Time Series Data: Pandas excels in handling time series data. It provides specialized functionality for working with time-based data, including date and time parsing, resampling, and time-based indexing.

4. Data Input and Output: Pandas supports reading and writing data in various formats, including CSV, Excel, SQL databases, and more. This makes it easy to import and export data from different sources.

5. Plotting and Visualization: Pandas provides built-in visualization capabilities that allow you to create basic plots and visualizations directly from your data.

# 4. Sympy

The [`sympy` package](https://docs.sympy.org/latest/index.html) can be used to perform symbolic mathematics calculations, such as exact/analytic integration, differentiation, finding limits, and many other mathematical operations. 

In the cell below we illustrate how to differentiate the equation 

$y = x\cos(x)$

NB: We expect the answer to be 
$\frac{dy}{dx} = \cos(x) - x\sin(x)$

```python
# import the library 
from sympy import *

# declare a symbols variable
# Note: 
# - This is ***NOT** a normal python variable; it is a sympy "symbol". 
# - These special symbolic variables are used within sympy to represent mathematical variables. 
x = symbols('x')

# differentiate the equation 
diff( x*cos(x), x)
```

# 5. Requests

The Python `requests` package is a popular third-party library that provides a simplified and convenient way to make HTTP requests and interact with web services, APIs, and websites. 

An API, or Application Programming Interface, facilitates communication between two pieces of software. It lets you retrieve and send data using code. We mostly commonly use APIs to retrieve data.

When we want to receive data from an API, we need to make a request. Requests are used all over the web. For instance, when you visit a website, your web browser made a request to the web server, which responds with the content of the web page. API requests work in exactly the same way – you make a request to an API server for data, and it responds to your request.

In the following we use the `get` method in the `requests` package to request information from the site called `api.open-notify.org`. This site happens to host a JSON file containing information on all astronauts who are in orbit right now, and the code below will download that JSON file.

```python
# Import the library 
import requests

# Define the web-address containing the data 
response = requests.get("http://api.open-notify.org/astros.json")

# Illustrate that a basic print of the response just gives a [200] message
# (this indicates success)
print(response)

# To get the data out of the response we use one of the following methods 
# (the one to use depends on what the specific website supplies: here it is a json)
# response.content() # Return the raw bytes of the data payload
# response.text()    # Return a string representation of the data payload
# response.json()    # This method is convenient when the API returns JSON
print("JSON response = ...")
print(response.json()) 
```
