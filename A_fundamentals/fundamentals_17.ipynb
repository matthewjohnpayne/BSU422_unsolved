# Fundamentals 17: Creating a Package

In Fundamentals 5, we learned how to save a function into a file and then import it as a *module*. 
 - A module in Python is a single file containing Python code, which can include functions, classes, variables, and statements. Modules allow you to break down your code into smaller, manageable files, making it easier to maintain and reuse.
 
In various notebooks we imported *packages* such as `numpy` and `matplotlib`. 
 - A package in Python is a collection of related modules organized within a directory hierarchy.
 
If you are simply making one or two functions to perform some analysis, a module is all you need to create. 
 - But if you are doing larger, more complex tasks, with lots of files and modules that perform a variety of interrelated tasks, it's often helpful to group them together into a package. 

In this notebook you will make a package of your own.



# 1. Package Structure 

A typical python package has the following structure 

```python
fundamentals_17_package/
    __init__.py
    data_generation.py
    data_analysis.py
    ```

Here we have ...
 - a package named `fundamentals_17_package` 
 - a couple of modules within it named `data_generation` and `data_analysis` 
 - a file named `__init__.py`$^*$ 
 
In the cells below, we will 
 - create these files
 - populate them with code  
 - import the package code 
 - use the code 

$^*$ N.B. It used to be absolutely *required* to have the `__init__.py` file in order to have a python package. These days it is *not* essential, but is regarded as good practice, and provides a few benefits ...
   - Initialization Code: You can include initialization code in the __init__.py file that gets executed when the package is imported. This can be useful for setting up resources, defining variables, or performing other setup tasks.
   - Explicitness: Including the __init__.py file explicitly indicates that the directory is intended to be a package. It enhances the clarity of your code and helps other developers understand your package's structure.
   - Compatibility: While recent versions of Python allow packages without an __init__.py file to be importable, older versions of Python may require the __init__.py file for package functionality to work as expected.
   - Third-Party Libraries: Many third-party tools and utilities (such as packaging tools and documentation generators) expect packages to have an __init__.py file.



# 2. Package Content 

Make the package directory

````python
# Make the package directory
import os 
os.mkdir('fundamentals_17_package')
````

Make a module named `__init__.py` within "fundamentals_17_package"
 - Here we use the `%%file` magic-command to generate the file 
 - Again, you can do this by hand if you prefer

````python
# Create an empty __init__.py file for the package "fundamentals_17_package" 
%%file fundamentals_17_package/__init__.py
````

Make a module named `data_generation.py` within "fundamentals_17_package"
 - This module contains a couple of functions (named `generate_random_veggie_list` & `generate_shopping_list` ) that generate data. 
 - Here we use the `%%file` magic-command to generate the file 

````python
%%file fundamentals_17_package/data_generation.py
"""
fundamentals_17_package/data_generation.py

A module containing various data-generation functions 
"""
import random 

def generate_random_veggie_list( n=5 ):
    """
    Generate a list of vegetables
    
    optional argument: n: integer: the total number of types of vegetable to be returned 
    
    return: veggie_list: list: list of vegetable names 
    """
    # names of veg to choose from 
    veg_names=["Tomatoes","Eggplant","Onions",
            "Potatoes","Cucumbers","Beets",
            "Scallions","Carrots","Artichokes",
            "Parsnips","Shallots","Garlic",
            "Peas","Spinach","Swiss Chard"]
    # use random.sample to choose n veggies
    return random.sample(veg_names, n)
    
def generate_shopping_list( veggie_list ):
    """
    Generate a dictionary of vegetables to be purchased
    
    required argument: veggie_list: list of vegetable names 
    
    return: veggie_dict: dictionary: key = vegetable-name, value = number of each to be purchased
    """
    veggie_dict = { veg:random.randint(1, 11) for veg in veggie_list } 
    return veggie_dict
````

 ### **Look on your file-system and check that `fundamentals_17_package/data_generation.py` got created with the expected content**

Make a module named `data_analysis.py` within "fundamentals_17_package"
 - This module contains a function (named `analyze_shopping` ) that analyzes data. 
 - Here we use the `%%file` magic-command to generate the file 

````python
%%file fundamentals_17_package/data_analysis.py
"""
fundamentals_17_package/data_analysis.py

A module containing various data-analysis functions 
"""
import random 

def analyze_shopping( list_of_veggie_dicts ):
    """
    Analyze multiple shopping lists
    
    required argument: list_of_veggie_dicts: lists: each item of the list is a *veggie_dict*
     - Each veggie_dict is expected to come from *generate_shopping_list* 
    
    return: mvp: tuple: most frequently purchased veggie, and the total amount purchased
    """
    # create a dict with cumulative counts ...
    total_dict = {} 
    for veggie_dict in list_of_veggie_dicts:
        for veg, amount in veggie_dict.items():
            if veg not in total_dict:
                total_dict[veg] = amount
            else:
                total_dict[veg] += amount
                
    # analyze overall stats 
    # NB: if there happen to be multiple matches I am lazily returning the first one ...
    max_value = max(total_dict.values())
    max_key = [key for key, value in total_dict.items() if value == max_value][0]
    
    return (max_key, max_value)
````

 ### **Look on your file-system and check that `fundamentals_17_package/data_analysis.py` got created with the expected content**

# 3. Package Import & Use

Now import and use the package ...

````python
# Import the "data_generation" & "data_analysis" packages from the "fundamentals_17_package" package
from fundamentals_17_package import data_generation, data_analysis

# Call the *generate_shopping_list* function multiple times to create a list-of-dictionaries 
list_of_dicts = [ data_generation.generate_shopping_list( data_generation.generate_random_veggie_list() ) for n in range(7) ] 

# Call the *analyze_shopping* function to analyze the data ...
result = data_analysis.analyze_shopping( list_of_dicts )
print(f'Most common purchase is {result[0]}: Number purchased = {result[1]}')
````
