## 2.1. Selecting Data


Let's use the `titanic` dataframe to learn how to do some basic data exploration using built-in methods from the `pandas` package

```python
titanic = pd.read_csv("https://raw.githubusercontent.com/pandas-dev/pandas/main/doc/data/titanic.csv")
```

## 2.1.1: Selecting by Column Label/Name

We saw above (from using the `info` and/or `head` functions) that one of the fields in the `titanic` dataframe is called "Age".

We can use the field-labels to select just the data in that one field.
 - In the example below we select the field just as we do in a dictionar.

The data that is returned is a list-like object which has a type `series`

```python
# Select the field named "Age"
s = titanic["Age"]

# Print out the information for the returned object
s.describe()
```

You can also select the data in a column as follows:
`s = titanic.Age`

I.e. the dataframe has a named "attribute", `Age`, that you can select

```python
s = titanic.Age
s.describe()
```

```python
# The `head` function also works on series (as well as the full dataframe)
s.head()
```

## 2.1.2: Selecting Rows: Slice Notation

We can use the slice notation from "fundamentals_4.ipynb" to select a subset of the rows from a dataframe 

```python
# Select rows with indicees 3-to-6 (inclusive)
rs = titanic[3:7]

# The returned object is a dataframe 
rs.info()
```

```python
# The returned object 
rs
```

## 2.1.3 Selecting Rows by Condition

pandas makes it very easy to select subsets of the rows in a dataframe according to the values in certain fields. 

As an example, in the cells below we will use the above `titanic` dataframe, and we will select all passengers over the age of 20. 

```python
# Create the dataframe 
titanic = pd.read_csv("https://raw.githubusercontent.com/pandas-dev/pandas/main/doc/data/titanic.csv")
# Create a *Series* of booleans (True/False) that labels whether or not the value in the `Age` field is > 20 
s = titanic['Age'] > 20
s
```

```python
# Use the Series, `s`, to select a subset of the rows in the dataframe 
above_twenty = titanic[s]
print(f'len(titanic) = {len(titanic)}')
print(f'len(above_twenty) = {len(above_twenty)}')
above_twenty
```

# You can combine the above steps that create a series and then select the rows into a *single step*

```python
# You can combine the creation of the series and the selection of the rows into a single step
above_twenty = titanic[titanic['Age'] > 20]
above_twenty
```

## 2.2. Basic Data Exploration : Sorting Data

The `DataFrame.sort_values()` method is used to sort a DataFrame by its column or row values. 

```python
titanic.head() 
```

```python
sorted_titanic = titanic.sort_values(by="Age")
sorted_titanic.head()
```

Note from the table above, that after using the `sort_values` command, the youngest passengers now appear at the top of the table. 

The `by` parameter can take a list of column names.
 - When a list of columns is supplied, the table is sorted first using the first field in the list, and then if any items in the first field are the same, the second field in the list is used as the basis for sorting. 


```python
sorted_titanic2 = titanic.sort_values(by=["Age","PassengerId"])
sorted_titanic2.head()
```
