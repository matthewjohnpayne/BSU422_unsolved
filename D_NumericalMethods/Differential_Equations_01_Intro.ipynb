## What is a differential equation ?

<div class="alert alert-block alert-info">
<b>A Differential Equation is an equation involving a function and one or more of its derivatives:</b> 
</div>

### Examples of differential equations

 1. $ \frac{dy}{dx} = 5y -3 $ 
  - $y$ is a function of x, e.g. $y=f(x)$
  - We can write this function as $ \frac{df(x)}{dx} = 5f(x) -3 $ 
 
 <br>
 
 1. $ y + \frac{dy}{dt} = 5t $ 
  - Here the independent variable is $t$ 

 <br>

 1. $ \frac{dx}{dt} = 7x^2t^3$ 
  - Here the independent variable is $t$ and the dependent variable is $x$ 

 <br>

 1. $ \frac{d^2y}{dx^2} + \frac{dy}{dx} + y = x $ 
  - Here the equation also involves the *second* derivative of $y$ w.r.t. $x$: $\frac{d^2y}{dx^2}$


# What are the solutions to a differential equation?

<div class="alert alert-block alert-info">
<b>The solution to a Differential Equation is a <u>function</u> (or a set of functions)</b> 
</div>

# An example solution

We want to solve the following equation:

$ \frac{dy}{dx} = 5y -3 $ 


#### Step 1: multiply by dx and divide by 5y−3:

$ \frac{dy}{5y -3} = dx $ 


#### Step 2: Integrate both sides of the equation :

$\begin{eqnarray}
 \int \frac{dy}{5y -3}   &=& \int dx \\
 \frac{1}{5}\log{|5y-3|} + K_1 &=& x + K_2 \\
 \frac{1}{5}\log{|5y-3|} &=& x + C\\
 \end{eqnarray}$
  - N.B. Constants of integration from both sides, $K_1$ & $K_2$, have been combined into $C$


#### Step 3: Rearrange :

$\begin{eqnarray}
5y-3 &=& \pm e^{[5(x + C)]} \\
y    &=& \pm \frac{1}{5}e^{[5(x + C)]} + \frac{3}{5} \\ 
y    &=& \pm \frac{1}{5}e^{[5(x + C)]} + \frac{3}{5} \\ 
y    &=& \pm \frac{1}{5}e^{5x}e^{5C} + \frac{3}{5} \\ 
y    &=& \pm Ke^{5x} + \frac{3}{5} \\ 
 \end{eqnarray}$
   - N.B. $K\equiv\frac{1}{5}e^{5C}$
   
#### Step 4: Check that the solution actually satisfies the original equation!:

Assuming 

$
\begin{eqnarray}
y    &=& \pm Ke^{5x} + \frac{3}{5} \\ 
\end{eqnarray}
$, 
 
then we find 


$\begin{eqnarray}
 \frac{dy}{dx}   &=& \pm 5Ke^{5x} \\
 & \&&\\
5y -3           &=& 5\left( \pm Ke^{5x} + \frac{3}{5}\right) - 3  &=& \pm5Ke^{5x} \\ 
 \therefore &&\\
 \frac{dy}{dx}   &=& 5y -3
 \end{eqnarray}
$


# Points to exphasize about the above example solution

<div class="alert alert-block alert-info">
<b>
    <ul> <li> There are two solutions (positive and negative) </li> </ul>	
    <ul> <li> The solutions are functions of x  </li> </ul>	
    <ul> <li> The solutions involve exponentials  </li> </ul>	
</b> 
</div>



 
The fact that the functions involve exponentials should make some kind of sense:
   - The differential eqn we started with, $ \frac{dy}{dx} = 5y -3 $, is saying that there is some function, $y$, that behaves like a shifted (by 3) and scaled (by 5) version of its own derivative.
   - I.e. the function and its derivative are kind of similar to one another. 
   - And we know that the exponential function, $y=e^x$, is equal to its own derivative: $dy/dx = e^x = y$ 
   - So it makes sense that the solution to the differential equation is some kind of exponential.

# Classification of Differential Equations 

## Ordinary -versus- Partial Differential Equations

### "Ordinary Differential Equations" (ODEs) 
<div class="alert alert-block alert-info">
<b>
    ODEs have a single independent variable.
</b> 
</div>
E.g.

$\frac{d^2y}{dx^2} + 2\frac{dy}{dx} -8 y = 0 $

where

$x$ is the independent variable, and 
$y$ is the dependent variable

### "Partial Differential Equations" (PDEs) 
<div class="alert alert-block alert-info">
<b>
    PDEs have two or more independent variables.
</b> 
</div>
E.g. 
$\frac{\partial^2 u}{\partial x \partial y} = 2x - y $

where the dependent variable, $u$, here is a function the two independent variables, $x$ and $y$.

$u(x, y)$

## Order of an Equation

<div class="alert alert-block alert-info">
<b>
    The order of a differential equation is the <u>highest</u> derivative contained in the equation.
</b> 
</div>
I.e. Is it a first derivative? A second derivative? etc:

#### An example of a *First Order* ODE
$ \frac{dy}{dx} = 5y -3 $ 

#### An example of a *Second Order* ODE
$\frac{d^2y}{dx^2} + 2\frac{dy}{dx} -8 y = 0 $

#### An example of a *Third Order* ODE
$\frac{d^3y}{dt^3} + \frac{d^2y}{dt^2} = t $



## Degree of an Equation

<div class="alert alert-block alert-info">
<b>
The degree is the exponent of the highest derivative.
</b> 
</div>

#### An example of a *First Degree* ODE
$\frac{d^3y}{dt^3} + \frac{d^2y}{dt^2} = t $

 - The highest derivative is $d^3y/dt^3$
 - The highest derivative has no explicit exponent, so it is a *First Degree* ODE
 - Considering the *Order* and the *Degree*, this is a *Third Order, First Degree* ODE

#### An example of a *Second Degree* ODE
$\left(\frac{d^2y}{dx^2}\right)^2 + \left(\frac{dy}{dx}\right)^3 -8 y = 0 $

 - The highest derivative is $d^2y/dx^2$
 - The highest derivative has an exponent of 2, so it is a *Second Degree* ODE
 - Considering the *Order* and the *Degree*, this is a *Second Order, Second Degree* ODE

#### An example of a *Third Degree* ODE
$\left(\frac{dy}{dt}\right)^3 + y = t^4 $

 - The highest derivative is $dy/dx$
 - The highest derivative has an exponent of 3, so it is a *Third Degree* ODE
 - Considering the *Order* and the *Degree*, this is a *First Order, Third Degree* ODE


## Linear -versus- Non-Linear Equations 

<div class="alert alert-block alert-info">
<b>
A differential equation is *linear* if the dependent variable (and its derivatives) have no exponent or other function put on it.
</b> 
</div>

I.e. If $y$ is the dependent variable, then a linear ODE cannot contain any exponents on $y$ such as $y^2$ or ..., and cannot contain any functions of $y$ such as $\sin(x)$, $\ln(x)$, etc.

N.B. Linear ODEs *can* contain functions of the *independent* variable.


#### Examples of Linear ODEs
$
\begin{eqnarray}
&& dy/dt = y \\
&& dy/dt = ty \\
&& d^2y/dx^2 + dy/dx = xy \\
&& dz/dx + x^2 + \sin(x) = 0 \\
\end{eqnarray}
$



#### Examples of Non-Linear ODEs
$
\begin{eqnarray}
&& dy/dt = y^2 \\
&& dx/dt + 1/x = 0 \\
&& d^2y/dx^2 + sin(y) = xy \\
&& z dz/dt = 0 \\
\end{eqnarray}
$
