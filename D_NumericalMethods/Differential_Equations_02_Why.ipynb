## The Importance of Differential Equations

Mathematical descriptions of changes use *differentials* and *derivatives*. 

Various differentials, derivatives, and functions become related via equations, such that a *differential equation* is a result that describes dynamically changing phenomena, evolution, and variation. 

Often, quantities are defined as the rate of change of other quantities (for example, derivatives of displacement with respect to time), or gradients of quantities, which is how they enter differential equations.

## Example 1: Rabbits

###  1.1 Analytic Solution 

 - In many areas of science (physics, biology, etc) we can often *measure* the current value of some physical quantity. 
  - E.g. $N$, the number of rabbits in a population
  
 - We often also have an equation describing how that physical quantity changes (perhaps as a function of time)
  - E.g. $\frac{dN}{dt} = rN$
    - (A linear, first-order, first-degree ODE)
  - Here someone has come up with a theory that suggests that the rate of change in the number of rabbits is proportional to the total size of the population.
    - I.e. perhaps they are assuming that a certain fraction of the rabbits have offspring every year. 
  
 - The problem that we then want to understand is how $N$ is going to evolve as a function of time, $N(t)$.
   - I.e. how many rabbits will there be as time goes on?
  
 - Solving the above integral analytically, we find
  - Rearranging & integrating ...
  > $\begin{eqnarray}
 \frac{dN}{dt}           &=& rN \\
 \int \frac{dN}{N}      &=& \int r dt \\ 
 \log{|N|}              &=& rt + C \\
 N                      &=& \pm e^{rt + C} \\ 
                        &=& \pm N_0 e^{rt} \\ 
\end{eqnarray}$

  - Checking ...
  >$\begin{eqnarray}
 \frac{dN}{dt}          &=& rN_0 e^{rt} = rN \\
\end{eqnarray}$

  - So we have exponential growth in the population of rabbits! 

###  1.2 Dynamic Plot of the Solution 

In the following example we plot the above solution to the growth in the population of rabbits.

We demonstrate how to use the `ipywidgets` `interact` module to make slider-bars to allow the user to dynamically control the plot

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact

def plot_func(init_popn, growth_rate):
    """ 
    Plotting function
     - Will be called using ipywidgets.interact 
     
    init_popn: # The initial population of rabbits at time t = 0
    growth_rate: # The equation parameter, $r$, that governs the rate of growth of the population
    
    """
    plt.figure(figsize=(8, 6))
    
    # An array of time values
    t = np.arange(0,15,0.01) 
    
    # Set up variables that will be controlled by sliders 
    N = init_popn*np.exp(growth_rate*t) # Using the equation we derived above to model the number of rabbits as a function of time

    # Plot the data 
    plt.plot(t, N,label="$N(t) = N_0 e^{rt}$")

    # Format the plot 
    plt.xlabel("t [years]")
    plt.ylabel("N [rabbits]")
    plt.xlim(t[0], t[-1] )
    plt.legend( loc=1 )
    #plt.ylim(-10, 10)
    plt.show()

# Call the plot function using the ipywidgets `interact` function
_ = interact(plot_func, init_popn=(1,50,1), growth_rate=(0., 1.,0.01))


```

## Example 2: Newton's Second Law applied to a Mass on a Spring

Another example concerns Newton's second law of motion: Here the relationship between the displacement x and the time t of an object under the force F is given by the differential equation

$ m{\frac {\mathrm {d} ^{2}x(t)}{\mathrm {d} t^{2}}}=F(x(t))$

which constrains the motion of a particle of constant mass m. 

In general, F is a function of the position x(t) of the particle at time t. 

The unknown function x(t) appears on both sides of the differential equation, and is indicated in the notation F(x(t)).


## Specific Example: Mass-on-a-Spring

Consider a mass on a spring as illustrate in the figure below ([link to original webpage]).(https://math.libretexts.org/Bookshelves/Calculus/Calculus_(OpenStax)/17%3A_Second-Order_Differential_Equations/17.03%3A_Applications_of_Second-Order_Differential_Equations). 

According to Hooke’s law, the restoring force of the spring is proportional to the displacement and acts in the opposite direction from the displacement, so the restoring force is given by  $−𝑘(𝑠+𝑥)$.

The spring constant, $k$, is given in pounds per foot in the English system and in newtons per meter in the metric system.

Now, by Newton’s second law, the sum of the forces on the system (gravity plus the restoring force) is equal to mass times acceleration, so we have

$m{\frac {\mathrm {d} ^{2}x(t)}{\mathrm {d} t^{2}}} = −𝑘(𝑠+𝑥)+𝑚𝑔 = −𝑘𝑠−𝑘𝑥+𝑚𝑔 $.
 
However, by the way we have defined our equilibrium position, $𝑚𝑔=𝑘𝑠$, the differential equation becomes
$ m{\frac {\mathrm {d} ^{2}x(t)}{\mathrm {d} t^{2}}} + kx = 0 $


![Mass on a Spring](MoS.png)


Solving ...    
    
$ m{\frac {\mathrm {d} ^{2}x(t)}{\mathrm {d} t^{2}}} + kx = 0 $

It turns out to be useful to define $\omega^2 = k^2/m^2$, giving ...

$ {\frac {\mathrm {d} ^{2}x(t)}{\mathrm {d} t^{2}}} + \omega^2 x = 0 $

In the previous notebook, [Differential_Equations_01_Intro.ipynb](Differential_Equations_01_Intro.ipynb), we saw that the general solution to this equation is 

$x =  (A+B)\cos{\left(\omega t\right)} - (A-B)i\sin{\left(\omega t\right)} $

**Since the equation of motion is real**, $A$ and $B$ should be complex conjugates of each other in order to cancel out the imaginary parts, making $x(t)$ a real function. 

Therefore, let $A=(C+Di)/2$ and $B=(C−Di)/2$, where $C$ and $D$ are real numbers. 

Now the equation becomes

$x(t) =  C\cos{\left(\omega t\right)} + D\sin{\left(\omega t\right)} $

If desired this can also be written in a *phase-shift* form, 

$x(t) = E \cos( \omega t + \delta)$
