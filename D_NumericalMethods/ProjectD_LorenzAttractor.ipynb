### Animations 

In this Lorenz Attractor project, it will be required that you make an *animation*. 

In the two cells below we demonstrate how to make a simple animation (of a sine-wave growing in length).

The animation you make in the project will be more complicated, but this simple demonstration should provide you with the ability to animate your own plots. 


```python
# Set up the figure for plotting 
fig = plt.figure()
plt.xlim(0, 2*np.pi)
plt.ylim(-1.1,1.1)

# Set up a `Camera` object to use to record image-frames 
camera = Camera(fig)     

# Create the individual images/frames 
for i in np.linspace(0, 2*np.pi, 128):
    x = np.arange(0,i,0.1)
    plt.plot(x,np.sin(x))
    camera.snap()

# Create the animation 
animation = camera.animate(interval = 20,   # <<-- Delay between frames in milliseconds.
                           repeat = False)

# Save the animation
filename = 'sine.mp4'
animation.save(filename)

# Show the animation 
Video(filename)
```

# 3D Animation ... 
The following cell is largely the same as above, but creates an animation for 3D data ...

```python
# Set up the figure for plotting 
fig = plt.figure()
ax = fig.add_axes([0, 0, 1, 1], projection='3d') # <<-- 3D plotting ... 
ax.set_xlim(0, 10*np.pi)
ax.set_ylim(-1.1,1.1)
ax.set_zlim(-1.1,1.1)

# Set up a `Camera` object to use to record image-frames 
camera = Camera(fig)

# Create the individual images/frames 
for i in np.linspace(0, 10*np.pi, 128):
    x = np.arange(0,i,0.1)
    plt.plot(x,np.sin(x),np.cos(x), color='blue')   # <<-- y=sin(x), z=cos(x)
    camera.snap()

# Create the animation 
animation = camera.animate(interval = 20,   # <<-- Delay between frames in milliseconds.
                           repeat = False)

# Save the animation
filename = 'sine3D.mp4'
animation.save(filename)

# Show the animation 
Video(filename)

```

## Lorenz Attractor: Background

The Lorenz system is a system of ordinary differential equations studied by mathematician and meteorologist Edward Lorenz. 

This set of differential equations has *chaotic solutions* for certain parameter values and initial conditions. 

The Lorenz equations can be represented as:

$\begin{aligned}
\frac{du}{dt} &= \sigma (v - u) \\
\frac{dv}{dt} &= \rho u - v - u w \\
\frac{dw}{dt} &= -\beta w + u v
\end{aligned}
$

Here, $\sigma$, $\beta$, and $\rho$ are the parameters for the system, and $(u, v, w )$ are the variables describing the state of the system at any given time $t$.

The original physical system that Edward Lorenz was investigating with this equation was a simple 2-dimensional model of a fluid in which:
 - $x$ is proportional to the rate of convection
 - $y$ is proportional to the horizontal temperature variation in the layer
 - $z$ is proportional to the vertical temperature variation; 
 - $\sigma$, $\beta$, and $\rho$ are parameters proportional to the Prandtl number, Rayleigh number, and the physical dimensions of the layer.

See the relevant [wikipedia article](https://en.wikipedia.org/wiki/Lorenz_system) for further details.  

In this problem, we will concentrate on the numerical evolution of the variables $(u, v, w )$ as a function of time, $t$, and "forget" about the physical interpretation of the values.
