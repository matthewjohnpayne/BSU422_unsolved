# Project A: Creating a Package

In [fundamentals_16.ipynb](../A_fundamentals/fundamentals_16.ipynb) we learned how to make a package.

In this project you will create a package with the following name and structure.

```python
project_a/
    __init__.py
    data.py
    analysis.py
    plotting.py
    ```

## Project Specification 

a. Following a similar approach to that described in [fundamentals_16.ipynb](../A_fundamentals/fundamentals_16.ipynb), create the `project_a` directory for your package, and the `__init__.py` file within it

b. In `data.py` create a function named `fetch( cityname ) ` that takes a `cityname` string variable as its single required arguement, and then 
 - uses the `requests` package from [Fundamentals 15](../A_fundamentals/fundamentals_15.ipynb) to read  historical temperature data from the `www.visualcrossing.com` website.
   - details on using the `www.visualcrossing.com` website & API are provided below. 
 - stores the returned data as a JSON file (using methods from [Fundamentals 14](../A_fundamentals/fundamentals_14.ipynb)).
 - returns the name of the JSON file
 
c. Write a `test_fetch` function that checks to see whether your `fetch` function works as expected. 

d. In `analysis.py`  create a function named `extract` that takes a `json_filename` string variable as its single required arguement, and then ...
 - reads a JSON file from the `json_filename` into a dictionary 
 - extracts the date and temperature information from the dictionary and puts them into *numpy arrays*
 - returns the numpy arrays
 
e. Write a `test_extract` function that checks to see whether your `extract` function works as expected. 

f. In `plotting.py`  create a function named `compare` that can take in two sets of numpy arrays (generated for two different cities using two different calls to `analysis.py`), and then plots
 - the temperature for the two cities as a function of time 
 - provides a legend, title and axis labels 
 - highlights the maximum temperature reached for each of the two cities
 
g. In a cell below, pull everything together by ...
 - Importing your package
 - Calling data.fetch() for two different cities
 - Calling analysis.extract() on the two JSON files from  data.fetch()
 - Calling plotting.compare() on the two sets of arrays from analysis.extract()

## Some details on API usage for `www.visualcrossing.com`
 - Set up an account
   - https://www.visualcrossing.com/weather-api
 - Get API-key
   - It will looks something like ... '4CBAT28U38NN1YHC5SWFB2H43'
 - Look at API documentation
   https://www.visualcrossing.com/resources/documentation/weather-api/timeline-weather-api/
 - Consider this sample query for the last 30 days of data for a to-be-specified location:
   https://weather.visualcrossing.com/VisualCrossingWebServices/rest/services/timeline/{location}/last30days?key={api_key}&include=days'
