# The Dice Game
 - Consider the game described in https://www.sonofacorner.com/an-introduction-to-monte-carlo/
 - I.e. you win $10 if the sum of 10 dice is in the range 40-50

## Simulating dice throws

Let's define a function that simulates the throw of ten dice, and then computes the sum of the results.

```python
# Original function from .../intro_monte_carlo/main_notebook.ipynb
def throw_and_sum_n_dice(n):
    '''
    This function throws n dice and sums the results
    '''
    total = 0
    for i in range(0,n):
        throw = randint(1,6)
        total += throw
    return total

# More efficient numpy version (that is possibly harder to understand)
def throw_and_sum_n_dice(n):
    '''
    This function throws n dice and sums the results
    The function "np.random.randint(...)" generates n numbers between 1 - 6 inclusive
    The function ""np.sum" sums-up all the randomly generated numbers
    '''
    return np.sum( np.random.randint(1, high=7, size=n) )
```

```python
n = 10 
print(f'The sum of {n} dice is: {throw_and_sum_n_dice(n)}')
```

## Simulating expected winnings 

Call the function a million times to check how often we would've won some money.


```python
# Original function from .../intro_monte_carlo/main_notebook.ipynb
def multiple_trials_n_dice_throws_and_sums(n_trials , n_throws=10):
    winning_occurances = 0
    for i in range(n_trials):
        total = throw_and_sum_n_dice(n_throws)
        if total >= 40 and total <= 50:
            winning_occurances += 1
    return winning_occurances
n_trials = 1000000
n_winning = multiple_trials_n_dice_throws_and_sums(n_trials , n_throws=10)

print(f'We would\'ve won money {n_winning:,.0f} times -- a.k.a. {n_winning/n_trials:.3%} of the time.') 
```

```python
# More efficient numpy version (that is  harder to understand)
def multiple_trials_n_dice_throws_and_sums(n_trials , n_throws=10):
    '''
    This function throws a dice "n_throws" and sums the results
     - The function "np.random.randint(...)" generates n numbers between 1 - 6 inclusive
     - The function ""np.sum" sums-up all the randomly generated numbers
     
    The above process is then repeated for a total of "n_trials" to see what the variation in 
    the total sum looks like 
    
    To select the winning trials, ...
     - The "np.nonzero" function is used to select the entries 40 - 50
     - "shape" is used to count the entries in the resultant array
    '''
    trials = np.sum( np.random.randint(1, high=7, size=(n_trials, n_throws) ), axis = 1 )
    return np.nonzero((40 <= trials) & (trials < 50))[0].shape[0] # <<-- N_winning

n_trials = 1000000
n_winning = multiple_trials_n_dice_throws_and_sums(n_trials , n_throws=10)


print(f'We would\'ve won money {n_winning:,.0f} times -- a.k.a. {n_winning/n_trials:.3%} of the time.') 
```

## Profit Path ...

Let's imagine that we now play the game fifty times and at the end we distribute the payouts.



```python
# Original function from .../intro_monte_carlo/main_notebook.ipynb
def compute_profit_path(n_iterations , n_throws=10):
    '''
    This function returns the profit (or loss) path at the end of 
    n_iterations iterations of the game, and where we throw n_throws dice.
    '''
    total_profit = 0
    for i in range(n_iterations):
        total = throw_and_sum_n_dice(n_throws)
        if total >= 40 and total <= 50:
            total_profit += 10
        else:
            total_profit += -2
    return total_profit

n_trials = 50
winnings = compute_profit_path(50)
print(f'On a single simulation of the game {n_trials} times we would {"win" if winnings > 0 else "lose"} {winnings} dollars')
```

```python
# More efficient numpy version (that is  harder to understand)
def compute_profit_path(n_iterations , n_throws=10):
    '''
    This function throws a dice "n_throws" and sums the results
     - The function "np.random.randint(...)" generates n numbers between 1 - 6 inclusive
     - The function ""np.sum" sums-up all the randomly generated numbers
     
    The above process is then repeated for a total of "n_iterations" 
    
    
    '''
    n_winning = multiple_trials_n_dice_throws_and_sums(n_iterations , n_throws=10)
    return 10*n_winning -2*(n_iterations-n_winning)
n_trials = 50
winnings = compute_profit_path(50)
print(f'On a single simulation of the game {n_trials} times we would {"win" if winnings > 0 else "lose"} {winnings} dollars')
```
