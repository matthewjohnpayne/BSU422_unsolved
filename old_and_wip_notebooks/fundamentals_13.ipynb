## 1. Sets
----------

A set is an unordered collection with no duplicate elements. 

Basic uses include membership testing and eliminating duplicate entries. 

Set objects also support mathematical operations like union, intersection, difference, and symmetric difference.

## 1.1 Defining Sets

Curly braces or the set() function can be used to create sets. 

Note: to create an empty set you have to use set(), not {}; the latter creates an empty dictionary, a data structure that we discuss in the next section.


```python
# Here we define a set named 'basket' using the curly-brackets approach
# When creating a set, any duplicates are removed
basket = {'apple', 'orange', 'apple', 'pear', 'orange', 'banana'}
print('basket=', basket)                      
```

```python
# Here we take an input list and turn it into a set using the set() approach
# Note that the set has fewer members because duplicates are not allowed into the set
bag_list = [ 'onion', 'potato' , 'carrot', 'radish', 'potato', 'onion', 'potato' ]
bag_set  = set( bag_list )
print('bag_list=', bag_list)                      
print('bag_set=', bag_set)                      
```

```python
# Note that when using the set() approach, we have to supply a single iterable as the argument:
# - We cannot supply multiple strings
#
# E.g., the following will fail, with an error, 
#          "TypeError: set expected at most 1 argument, got 7"
veg_set = set( 'onion', 'potato' , 'carrot', 'radish', 'potato', 'onion', 'potato' )
```

```python
# However, note that a single string *IS* an iterable (of multiple single characters).
# Hence if you pass a single string to set(), it will be interpreted as a set of individual letters
# (and any duplicates will be removed)
letter_set = set( 'knjdfvnfdkvnakjahbdsfaljahdsf' )
print(letter_set)
```

```python
# The elements in a set can be of different types
# NB: the elements can even be composite objects such as tuples 
x = {42, 'foo', 3.14159, None, ('v','bb',3) }
print(x)
```

## 1.2 Properties of Sets and the elements of Sets

```python
# We *can* add new elements to a set after the set has been created
x.add( 'dog' )
print(x)
```

```python
# But we can only add elements that are themselves immutable (unchangeable) 
# This means that we cannot make lists or dictionaries be a part of a set
#
# E.g. the following will fail with the error
#      TypeError: unhashable type: 'list'
x.add( ['cat','hamster'] )
```

## 1.3 Membership Testing

Sets can be used to perform logical testing to see whether a given item is in a set

```python
# Test whether the string 'orange' is in the set named *basket* 
print( 'orange' in basket )              

# Test whether the string 'crabgrass' is in the set named *basket* 
print('crabgrass' in basket)
```

Testing the membership of a set is quicker than testing the membership of a list 

```python
# Let's perform the same task 10 million times 
# We will see what the time difference is to check for membership in a list or in a set

import time

n_loops = 10000000

# Time how long it takes to check in a list 
start_time = time.time()
for n in range(n_loops):
    'orange' in bag_list
    'potato' in bag_list
end_time = time.time()
time_taken = end_time - start_time 
print('Number of seconds taken to check list:', time_taken)

# Time how long it takes to check in a set 
start_time = time.time()
for n in range(n_loops):
    'orange' in bag_set
    'potato' in bag_set
end_time = time.time()
time_taken = end_time - start_time 
print('Number of seconds taken to check set:', time_taken)
```

## 1.4 Mathematical Set Operations

Sets can be used to perform mathematical logical operations, including union, intersection, difference, and symmetric difference.

```python
# Define some sets of letters 
# NB: Each set will contain only unique letters, but there is some overlap between the two sets ...
set_a = set('hvtfvctbhvfhhvcthvbggvr')
set_b = set('jhgnyujnnhjuggnjunbj')
print("set_a = ", set_a) 
print("set_b = ", set_b)                       

# Find the letters that are in 'a' only and that are not in 'b'
print( "DIFFERENCE:", set_a - set_b )                         

# Find the unique letters that are the combination of 'a' and 'b'
print( "UNION:", set_a | set_b )                              

# Find the letters that are found in both 'a' and 'b'
print( "INTERSECTION:", set_a & set_b )                           

# Find the letters in 'a' or 'b' but not both
print( "SYMMETRIC DIFFERENCE:", set_a ^ set_b )                          
```

```python
# Define some sets of letters 
set_c = set('abc')
set_d = set('abcd')
set_e = {'e'}
set_f = set('abcd')
set_g = {'ab'}

print("set_c = ", set_c) 
print("set_d = ", set_d)  
print("set_e = ", set_e)  
print("set_f = ", set_f)  
print("set_g = ", set_g)  

# Find whether set_c is a subset of set_d
print( "c SUBSET of d? :", set_c <= set_d )     

# Find whether set_e is a subset of set_d
print( "e SUBSET of d? :", set_e <= set_d )                         

# Find whether set_f is a subset of set_d
print( "f SUBSET of d? :", set_f <= set_d )                         

# Find whether set_f is a PROPER subset of set_d
print( "f PROPER SUBSET of d? :", set_f < set_d )                         

# Find whether set_g is a subset of set_d
print( "g SUBSET of d? :", set_g <= set_d )                         
```

## 2. Dictionaries
----------
Python dictionaries are a powerful and versatile data structure used to store and manage collections of key-value pairs. 

Each key in a dictionary is unique and associated with a value. 

Dictionaries are often referred to as "associative arrays" or "hash maps" in other programming languages. 

Below is an introduction to dictionaries with some examples.

## 2.1 Creating a Dictionary:
You can create a dictionary by enclosing a comma-separated list of key-value pairs within curly brackets {}.

```python
# Creating a dictionary
student = {
    "name": "Munazza",
    "age": 20,
    "major": "Computer Science"
}
print(student)
```

## 2.2 Accessing Values:
You can access the values in a dictionary by referencing the keys

```python
# Accessing values
print(student["name"])    # Output: Munazza
print(student["age"])     # Output: 20
print(student["major"])   # Output: Computer Science
```

## 2.3 Adding and Modifying Entries:
You can add new entries to a dictionary or modify existing ones using the keys.

```python
# Adding a new entry
student["gender"] = "Female"
print(student)

# Modifying an entry
student["age"] = 21
print(student)
```

## 2.4 Dictionary Methods:
Dictionaries have several methods that allow you to manipulate and retrieve information. The [python documentation](https://docs.python.org/3/library/stdtypes.html#dictionary-view-objects) has a complete list of the methods available for dictionaries.

```python
# Getting all keys and values
keys = student.keys()
values = student.values()
print('keys=', keys)
print('values=', values)

# Checking if a key exists
if "name" in student:
    print("Name exists in the dictionary")

# Removing an entry
removed_value = student.pop("age")
print(student)
```

## 2.5 Iterating Through a Dictionary:
You can iterate through the keys or key-value pairs of a dictionary.

```python
# Iterating through keys
# (NB writing "for key in student" gives identical results to ""for key in student.keys() )
print('\n Iterating through keys ...')
for key in student:
    print(key, student[key])

# Iterating through key-value pairs
print('\n Iterating through key-value pairs ...')
for key, value in student.items():
    print(key, value)
```

## 2.6 Use Cases:
Dictionaries are handy in various scenarios, such as:
1. Mapping: Creating relationships between keys and corresponding values, such as:
    - Data Storage: Storing data with meaningful labels, like storing student information.
    - Configuration Settings: Storing configuration settings with associated values.
2. Counting Occurrences: Counting the occurrences of elements in a dataset. See example below.

```python
word_counts = {}
text = "This is The sample text to count word occurrences. It contains multiple words. Some of the words occur multiple times within the text."
words = text.split()

for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

for key in sorted(list(word_counts.keys())):
    print(f'The word "{key}" occurs {word_counts[key]} times')
```
